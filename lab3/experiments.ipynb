### Winning ratio with same search depths

##### Max first

```python
from play import play


pl1_win = 0
pl2_win = 0
draw = 0
iter = 50

for i in range (iter):
    winner = play(4,4, True)

    if winner == None:
        draw += 1

    elif winner.char == "1":
        pl1_win += 1

    else:
        pl2_win += 1


print("Max was first")
print(f'Player1: {(pl1_win/iter)*100}%')
print(f'Player2: {(pl2_win/iter)*100}%')
print(f'Draw: {(draw/iter)*100}%')
```

##### Min first

```python
from play import play


pl1_win = 0
pl2_win = 0
draw = 0
iter = 100

for i in range (iter):
    winner = play(3,3, False)

    if winner == None:
        draw += 1

    elif winner.char == "1":
        pl1_win += 1

    else:
        pl2_win += 1


print("Min was first")
print(f'Player1: {(pl1_win/iter)*100}%')
print(f'Player2: {(pl2_win/iter)*100}%')
print(f'Draw: {(draw/iter)*100}%')
```

#### Random moves

```python
import random
from two_player_games.player import Player
from two_player_games.games.connect_four import ConnectFourMove, ConnectFour, ConnectFourState

pl1_win = 0
pl2_win = 0
draw = 0
iter = 1000

for i in range(iter):
    player1 = Player("1")
    player2 = Player("2")
    game = ConnectFour((7,6), player1, player2)


    while not game.is_finished():

        index = random.randint(0,6)
        move = ConnectFourMove(index)
        game.make_move(move)


    winner = game.get_winner()

    if winner == None:
        draw += 1

    elif winner.char == "1":
        pl1_win += 1

    else:
        pl2_win += 1

print(f'Player1: {(pl1_win/iter)*100}%')
print(f'Player2: {(pl2_win/iter)*100}%')
print(f'Draw: {(draw/iter)*100}%')
```
