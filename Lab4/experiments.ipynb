#### Preparing data


```python
import pandas as pd

sample = 50

set = pd.read_csv('lab-4-dataset.csv')

set_shuffled = set.sample(frac=1, random_state=42).reset_index(drop=True)


train_ratio = 0.7
train_size = int(len(set) * train_ratio)

training_set = set_shuffled[:train_size]
validation_set_all = set_shuffled[train_size:]

validation_set = validation_set_all.sample(n=sample)

classified_value = training_set['HeartDisease']

training_features = (training_set["Sex"],training_set["ChestPainType"],training_set["FastingBS"],training_set["RestingECG"],training_set["ExerciseAngina"],training_set["ST_Slope"],training_set["AgeGroup"],training_set["RestingBP_Category"],training_set["Cholesterol_Category"],training_set["MaxHR_Category"],training_set["Oldpeak_Category"])

y_true = validation_set['HeartDisease'].to_numpy()


validation_set =  validation_set.drop(validation_set.columns[0], axis=1)
validation_features = validation_set.drop(columns=['HeartDisease'])

data_samples = [tuple(row) for row in validation_set.values]
```

#### Classification

```python
from random_forest import random_forest
import numpy as np



y_pred = np.zeros(sample)
probability = np.zeros(sample)


for i in range (sample):
    sample_features = data_samples[i]
    y, p = random_forest(training_features, classified_value, 3, 100, 12, sample_features)
    y_pred[i] = y
    probability[i] = p


accuracy = 0
precision = 0
recall = 0

for i in range(len(y_pred)):
    if y_pred[i] == y_true[i]:
        accuracy += 1
        if y_pred[i] == 1 and y_true[i] == 1:
            precision += 1
            recall +=1

y_pred_positive = (y_pred == 1).sum()
y_true_positive = (y_true == 1).sum()


print(f"accuracy: {(accuracy/len(y_pred)) * 100}%")

try:
    print(f"precision: {(precision/y_pred_positive) * 100}%")
    print(f"recall: {(recall/y_true_positive) * 100}%")

except ZeroDivisionError:
    print("No positive values")
```

```python
from sklearn.metrics import roc_curve, auc
import matplotlib.pyplot as plt


fpr, tpr, thresholds = roc_curve(y_true, probability)


roc_auc = auc(fpr, tpr)


plt.figure()
plt.plot(fpr, tpr, color='darkorange', lw=2, label=f'ROC curve (AUC = {roc_auc:.2f})')
plt.xlim([0.0, 1.0])
plt.ylim([0.0, 1.05])
plt.xlabel('False Positive Rate')
plt.ylabel('True Positive Rate')
plt.title('Receiver Operating Characteristic (ROC)')
plt.legend(loc="lower right")
plt.grid()
plt.show()
```

#### Comparing to Scikit-learn

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, precision_score, recall_score, roc_auc_score, roc_curve
import pandas as pd

df = pd.read_csv("lab-4-dataset.csv")

X = df.drop(columns=["HeartDisease"])
y = df["HeartDisease"]

X = pd.get_dummies(X)

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42, stratify=y)

model = RandomForestClassifier(n_estimators=150, max_depth=10, random_state=42)
model.fit(X_train, y_train)

y_pred = model.predict(X_test)
y_proba = model.predict_proba(X_test)[:, 1]

print("Accuracy:", accuracy_score(y_test, y_pred)*100,"%")
print("Precision:", precision_score(y_test, y_pred)*100,"%")
print("Recall:", recall_score(y_test, y_pred)*100,"%")
print("AUC:", roc_auc_score(y_test, y_proba))

```
