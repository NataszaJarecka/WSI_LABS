#### Algorithm's path

```python
import numpy as np
from genetic_algorithm import generate_population, roulette_selection, find_the_best


def genetic_algorithm_data(evaluation, gene_count, size, population_count, mutation_prob, generations):

    i = 1
    population = generate_population(gene_count, population_count, mutation_prob)

    best_individuals = []
    average_rate = []
    iter = []
    while i <= generations:
        new_population = np.empty(population_count, dtype=object)

        for k in range(population_count // 2):
            parent1 = roulette_selection(population, evaluation, size)
            parent2 = roulette_selection(population, evaluation, size)
            child1, child2 = parent1.reproduce(parent2)

            child1.mutate()
            child2.mutate()

            new_population[2 * k] = child1
            new_population[2 * k + 1] = child2

        if population_count % 2 != 0:
            parent1 = roulette_selection(population, evaluation, size)
            parent2 = roulette_selection(population, evaluation, size)
            child1, _ = parent1.reproduce(parent2)
            child1.mutate()
            new_population[-1] = child1

        best = find_the_best(population, evaluation, size)
        best_individuals.append(evaluation(best.get_genotype(), size))

        population_rate = np.array([evaluation(ind.get_genotype(), size) for ind in population])
        average_rate.append(sum(population_rate/population_count))

        iter.append(i)

        population = new_population
        i += 1


    return average_rate, best_individuals, iter


rate3, best3, iter = genetic_algorithm_data(EVALUATION, GENE_COUNT, SIZE, 700, 0.005, 1000)

plt.plot(iter, rate3, color='g', label='Średnia ocena osobników')

plt.xlabel('Generacja')
plt.ylabel('Ocena')
plt.title('Postęp algorytmu genetycznego')

plt.legend()
plt.grid()

plt.show()

```
