```python
from genetic_algorithm import genetic_algorithm
from evaluation import evaluate

evaluation = evaluate
gene_count = 400
size = 20
population_count = 80
mutation_prob = 0.1
generations = 80

best_solution = genetic_algorithm(evaluation, gene_count, size, population_count, mutation_prob, generations)

print(evaluate(best_solution.get_genotype()))
```

```python
import matplotlib.pyplot as plt

population_count = 50
iter = 50
x = []
y = []
for i in range(30):
    best_solution = genetic_algorithm(evaluation, gene_count, size, population_count, mutation_prob, iter)
    best_rate = evaluate(best_solution.get_genotype())
    x.append(i)
    y.append(best_rate)


plt.scatter(x, y, s=10, c='red')

plt.title('Wykres Punktowy')
plt.xlabel('Z')
plt.ylabel('Y')

plt.show()
```
